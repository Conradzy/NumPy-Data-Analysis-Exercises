```python
import numpy as np

arr = np.arange(1, 21)

sum_val = np.sum(arr)

mean_val = np.mean(arr)

multiplied = arr * 3

print("Array:", arr)
print("Sum:", sum_val)
print("Mean:", mean_val)
print("Multiplied by 3:", multiplied)
```

```python
import numpy as np

A = np.array([[2, 5],
              [7, 35]])

B = np.array([[6, 10],
              [7, 8]])

adding = A + B


subtracting = A - B

multiply = np.dot(A, B)

print("Addition:\n", adding)
print("Subtraction:\n", subtracting)
print("Matrix Multiplication:\n", multiply)
```

```python
import numpy as np

arr = np.arange(1, 21)
even_numbers = arr[arr % 2 == 0]

slice_arr = arr[5:16]

modified_arr = arr.copy()
modified_arr[modified_arr > 10] = 0

print("Original Array:", arr)
print("Even Numbers:", even_numbers)
print("Index 5 to 15:", slice_arr)
print("Modified Array:", modified_arr)
```

```python
import numpy as np

rand_arr = np.random.randint(1, 101, (5, 5))

max_val = np.max(rand_arr)

min_val = np.min(rand_arr)

std_val = np.std(rand_arr)

print("Random Array:\n", rand_arr)
print("Maximum:", max_val)
print("Minimum:", min_val)
print("Standard Deviation:", std_val)
```

```python
import numpy as np

data = np.array([
    [2, 65, 45],
    [4, 70, 50],
    [6, 80, 65],
    [8, 90, 85],
    [5, 75, 60],
    [np.nan, 85, 70]
])

print("Shape:", data.shape)

missing_values = np.isnan(data).sum(axis=0)
print("Missing values per column:", missing_values)

col_means = np.nanmean(data, axis=0)


indices = np.where(np.isnan(data))


data[indices] = np.take(col_means, indices[1])

print("Cleaned Data:\n", data)

mean_vals = np.mean(data, axis=0)

std_vals = np.std(data, axis=0)

correlation = np.corrcoef(data.T)

print("Mean:", mean_vals)
print("Standard Deviation:", std_vals)
print("Correlation Matrix:\n", correlation)

avg_study = np.mean(data[:, 0])

filtered_students = data[
    (data[:, 0] > avg_study) &
    (data[:, 1] > 75)
]

print("Students meeting criteria:\n", filtered_students)
```
